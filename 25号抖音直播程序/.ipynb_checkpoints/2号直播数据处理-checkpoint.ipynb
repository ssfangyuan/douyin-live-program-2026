```python
import pandas as pd
# df = pd.read_excel("./订单.xlsx")
# df = pd.read_csv("./订单.csv",encoding='gbk')
df =pd.read_csv("./1.csv",encoding='gbk')
```

```python
# df["订单提交时间"]
```

```python
# if '\t' in df.loc[0,'商家编码']:
#     df['商家编码'] = df2['商家编码'].str[:19]
if '\t' in df.loc[0,'订单提交时间']:
    df['订单提交时间'] = df['订单提交时间'].str[1:]
```

```python
from datetime import datetime
df['商家编码'] = df['商家编码'].str[1:]
df["订单提交时间"] = pd.to_datetime(df["订单提交时间"],format = "%Y-%m-%d %H:%M:%S")
df["订单提交时间"] = df["订单提交时间"].map(lambda x: x.strftime('%Y-%m-%d'))
df["销售金额"]= (df['商品数量']  * df['商品单价']) + (df['运费'])
df1=pd.read_excel("商品编码.xlsx")
df1.rename(columns={"商品名":"产品名称","商品编码":"商家编码"},inplace=True)
df1 = df1[["商家编码","产品名称"]]
df= pd.merge(df,df1,on="商家编码",how="left")
#df[df['产品名称'].isnull()]['商家编码'].value_counts()
# df = df[df['产品名称'] != '排扣收腹裤']
# df = df[df['产品名称'] != '睫毛膏']
# df = df[df['产品名称'] != 'V领蕾丝']
# df= df[df['广告渠道'] !=  '短视频']
```

```python
df = df.reset_index(drop=True)
# df
```

```python
df['广告渠道'].value_counts()
```

```python
df2 = pd.read_csv("./2.csv",encoding='utf-8')
# df2 = pd.read_csv("./3.csv")
# df2 = pd.read_csv("./异常运单数.csv",encoding='gbk')
# df2 = pd.read_excel("./异常运单数.xlsx")
df2.rename(columns={"订单号":"主订单编号"},inplace= True)
df['主订单编号'] = df['主订单编号'].astype(str)
df2['主订单编号'] = df2['主订单编号'].astype(str)
df2 = df2[["主订单编号","包裹状态","揽件时间"]]
df2["包裹状态"].replace('已签收','已完成1',inplace= True)
# df2["包裹状态"].replace('已拒收','已完成1',inplace= True)
df2["包裹状态"].replace('派送中','已完成1',inplace= True)
df2["包裹状态"].replace('待取件','已完成1',inplace= True)
df2 = df2.drop_duplicates(subset='主订单编号', keep='first') #删除重复项
if '\t' in df2.loc[0,'主订单编号']:
    df2['主订单编号'] = df2['主订单编号'].str[:19]
```

```python
# df2
```

```python
# df2 = pd.read_excel("./异常运单数 - 副本.xlsx")
```

```python
df3 = pd.merge(df,df2,on="主订单编号",how="left")
df3['主订单编号'] = df3['主订单编号'].astype(str)
df3['子订单编号'] = df3['子订单编号'].astype(str)
# df3 = df3.dropna(subset=['包裹状态'])
# df3.reset_index(drop=True,inplace= True)
for i in range (0,len(df3)):
    if df3.loc[i,"包裹状态"] == "已完成1" and (df3.loc[i,'售后状态'] == '-' or df3.loc[i,'售后状态'] == '售后关闭' or df3.loc[i,'售后状态'] == '换货成功'):
        df3.loc[i,"订单状态"] = "已完成"
```

```python
# 这一步可以排除取消支付、待发货、停发备货的订单
df3 = df3.dropna(subset=['包裹状态'])
df3.reset_index(drop=True,inplace= True)
```

```python
df3['订单数'] = df3['商品数量']
```

```python
# df3 = df3.dropna(subset=['包裹状态'])
df3["已发货"] = 0
df3["在线支付已完成"] = 0
df3["售后中"] = 0
df3["退回成功"] = 0
for i in range(0,len(df3)):
    if df3.loc[i,"订单状态"] == "已完成" and (df3.loc[i,"售后状态"] == "换货成功" or df3.loc[i,"售后状态"] == '-' or df3.loc[i,"售后状态"] == '售后关闭' or df3.loc[i,"售后状态"] == '补寄/换货成功'):
        df3.loc[i,"在线支付已完成"] = df3.loc[i,'订单数']

    elif df3.loc[i,'售后状态'] == '同意退款，退款成功':
        df3.loc[i,"退回成功"] = df3.loc[i,'订单数']

    elif df3.loc[i,"订单状态"] == '已发货' and (df3.loc[i,"售后状态"] == "换货成功" or df3.loc[i,"售后状态"] == '-' or df3.loc[i,"售后状态"] == '售后关闭' or df3.loc[i,"售后状态"] == '补寄/换货成功'):
        df3.loc[i,"已发货"] = df3.loc[i,'订单数']
        
    elif df3.loc[i,'售后状态'] != "-" and df3.loc[i,'售后状态'] != "换货成功" and df3.loc[i,'售后状态'] != "售后关闭" and df3.loc[i,'售后状态'] != "同意退款，退款成功":
        df3.loc[i,"售后中"] = df3.loc[i,'订单数']
```

```python
df3['在线支付已完成'].value_counts()
```

```python
df3['广告渠道'].value_counts()
```

```python
df3['复购']= 0
df3['直播'] = 0
df3['广告'] = 0

df3.loc[df3["广告渠道"] == '无','复购'] =1
df3.loc[df3["广告渠道"] == '直播','直播'] =1
df3.loc[df3["广告渠道"] == '商品卡','直播'] =1
df3.loc[df3["广告渠道"] == '短视频','广告'] =1
```

```python
df3.to_excel('包状态.xlsx')
```

```python
#综合订单表、编码表、异常运单表后做透视表
import numpy as np
df3['销售金额'].fillna(0, inplace=True)
df3= df3[df3['广告渠道'] !=  '短视频']
result = pd.pivot_table(df3,index=['订单提交时间'] , values = ['销售金额','订单数','在线支付已完成','售后中','退回成功','已发货','复购','直播','广告'] , aggfunc = np.sum)
result = result.reset_index()
result = result[['订单提交时间','订单数','销售金额','售后中','在线支付已完成','已发货','退回成功','复购','直播','广告']]
```

```python
#拼接店铺消耗表
df_consume = pd.read_excel("店铺消耗.xlsx")
df_consume = df_consume[['日期','产品','消耗','转化数']]
temp = df_consume.日期.str.extract('(\d+).*?(\d+).*?(\d+)') 
df_consume['订单提交时间'] = temp[0] + "-" + temp[1] + '-' + temp[2]
del df_consume['日期']
df_consume = pd.pivot_table(df_consume,index=['订单提交时间'] , values = ['消耗','转化数'] , aggfunc = np.sum)
df_consume = df_consume.reset_index()
df_consume["订单提交时间"] = pd.to_datetime(df_consume["订单提交时间"],format = "%Y-%m-%d")
df_consume["订单提交时间"] = df_consume["订单提交时间"].map(lambda x: x.strftime('%Y-%m-%d'))
```

```python
final = pd.merge(result,df_consume,on = ['订单提交时间'],how = 'left')
final = final[['订单提交时间','转化数','订单数','在线支付已完成','已发货','售后中','复购','退回成功','消耗','销售金额']]
final['真实发货率'] = final['订单数'] / final['转化数'] 
final['签收率'] = final['在线支付已完成'] / final['转化数'] 
final['退回率'] = final['退回成功'] / final['转化数'] 
final['转化成本'] = final['消耗'] / final['转化数']
final['平均客单'] = final['销售金额'] / final['订单数']
final['复购率'] = final['复购'] / final['转化数']
final['真实发货率'] = final['真实发货率'].apply(lambda x: format(x, '.2%'))
final['签收率'] = final['签收率'].apply(lambda x: format(x, '.2%'))
final['退回率'] = final['退回率'].apply(lambda x: format(x, '.2%'))
final['复购率'] = final['复购率'].apply(lambda x: format(x, '.2%'))
final['转化成本'] = final['转化成本'].round(decimals=2)
final['平均客单'] = final['平均客单'].round(decimals=2)
final.rename(columns={"转化数":"广告单量","订单数":"总发出单量","订单提交时间":"日期",'销售金额':'客单'},inplace=True)
final['产品'] = '直播产品'
final['备注'] = ' '
final= final[['日期','产品','广告单量','总发出单量','真实发货率','在线支付已完成','已发货','售后中','退回成功','签收率','退回率','消耗','平均客单',
             '复购率','备注','客单','复购']]
```

```python
# 生成广告表
df_advert = df3[df3['广告渠道'] != 
               '无']
result_advert = pd.pivot_table(df_advert,index=['订单提交时间'] , values = ['销售金额','订单数','在线支付已完成','售后中','退回成功','已发货','复购','直播','广告'] , aggfunc = np.sum)
final_advert = pd.merge(result_advert,df_consume,on = ['订单提交时间'],how = 'left')
final_advert = final_advert[['订单提交时间','转化数','订单数','在线支付已完成','已发货','售后中','退回成功','消耗','销售金额','复购','直播','广告']]
final_advert['真实发货率'] = final_advert['订单数'] / final_advert['转化数'] 
final_advert['签收率'] = final_advert['在线支付已完成'] / final_advert['转化数'] 
final_advert['退回率'] = final_advert['退回成功'] / final_advert['转化数'] 
final_advert['转化成本'] = final_advert['消耗'] / final_advert['转化数']
final_advert['平均客单'] = final_advert['销售金额'] / final_advert['订单数']
final_advert['真实发货率'] = final_advert['真实发货率'].apply(lambda x: format(x, '.2%'))
final_advert['签收率'] = final_advert['签收率'].apply(lambda x: format(x, '.2%'))
final_advert['退回率'] = final_advert['退回率'].apply(lambda x: format(x, '.2%'))
final_advert['转化成本'] = final_advert['转化成本'].round(decimals=2)
final_advert['平均客单'] = final_advert['平均客单'].round(decimals=2)
final_advert.rename(columns={"转化数":"广告单量","订单数":"总发出单量","订单提交时间":"日期",'销售金额':'客单'},inplace=True)
final_advert['产品'] = '直播产品'
final_advert['备注'] = ''
final_advert= final_advert[['日期','产品','广告单量','总发出单量','真实发货率','在线支付已完成','已发货','售后中','退回成功','签收率','退回率','消耗','平均客单',
             '备注','客单']]
```

```python
df_repeat = df3[df3['广告渠道'] == 
               '无']
result_repeta = pd.pivot_table(df_repeat,index=['订单提交时间'] , values = ['销售金额','订单数','在线支付已完成','售后中','退回成功','已发货','复购','直播','广告'] , aggfunc = np.sum)
final_repeat = pd.merge(result_repeta,df_consume,on = ['订单提交时间'],how = 'left')
final_repeat = final_repeat[['订单提交时间','转化数','订单数','在线支付已完成','已发货','售后中','退回成功','消耗','销售金额','复购','直播','广告']]
# final_repeat = pd.pivot_table(df_repeat,index=['订单提交时间','产品名称'] , values = ['销售金额','订单数','在线支付已完成','售后中','退回成功','已发货','复购','直播','广告'] , aggfunc = np.sum)
final_repeat['复购发货率'] = final_repeat['订单数'] / final_repeat['转化数'] 
final_repeat['签收率'] = final_repeat['在线支付已完成'] / final_repeat['转化数'] 
final_repeat['退回率'] = final_repeat['退回成功'] / final_repeat['转化数'] 
final_repeat['转化成本'] = final_repeat['消耗'] / final_repeat['转化数']
final_repeat['平均客单'] = final_repeat['销售金额'] / final_repeat['订单数']
final_repeat['复购发货率'] = final_repeat['复购发货率'].apply(lambda x: format(x, '.2%'))
final_repeat['签收率'] = final_repeat['签收率'].apply(lambda x: format(x, '.2%'))
final_repeat['退回率'] = final_repeat['退回率'].apply(lambda x: format(x, '.2%'))
final_repeat['转化成本'] = final_repeat['转化成本'].round(decimals=2)
final_repeat['平均客单'] = final_repeat['平均客单'].round(decimals=2)
# final['平均客单'] = final['平均客单'].astype(float)
final_repeat.rename(columns={"转化数":"广告单量","订单数":"复购发出单量","订单提交时间":"日期",'销售金额':'客单'},inplace=True)
final_repeat= final_repeat[['日期','广告单量','复购发出单量','复购发货率','在线支付已完成','已发货','售后中','退回成功','签收率','退回率','消耗','转化成本','平均客单',
             '客单','复购']]
final_repeat['产品'] = '直播产品'
final_repeat['备注'] = ''
final_repeat= final_repeat[['日期','产品','广告单量','复购发出单量','复购发货率','在线支付已完成','已发货','售后中','退回成功','签收率','退回率','平均客单',
             '备注','客单']]
```

```python
writer = pd.ExcelWriter('直播汇总.xlsx')
final.to_excel(writer, sheet_name='汇总', index=False)
final_advert.to_excel(writer, sheet_name='广告', index=False)
final_repeat.to_excel(writer, sheet_name='复购', index=False)
writer.save()
```
