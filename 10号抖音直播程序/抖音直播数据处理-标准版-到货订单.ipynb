```python
# 订单数会rename为总发出单量
df3['订单数'] = df3['商品数量']
```

```python
# 需要清除未支付订单：①取消原因：非空②包裹状态：空 并且 订单状态：关闭 并且 售后状态：-
# 排除 (订单状态：已关闭 且 售后状态：- ) 即可实现前一点
df3.drop(df3[(df3.订单状态 == '已关闭') & (df3.售后状态 == '-')].index, inplace=True) 
df3.reset_index(drop=True,inplace= True)
df3['总支付单量'] = df3['订单数']
df3['售前退款单量'] = df3[(df3['包裹状态'].isnull())&(df3['售后状态']=='退款成功')]['订单数']
df3['揽件前退款单量'] = df3[(df3['揽件时间'].isnull())&(df3['售后状态']=='退款成功')]['订单数']
df3['备货单量'] = df3[(df3['包裹状态'].isnull())&(df3['订单状态']=='待发货')]['订单数']

df3['售前退款单量'].fillna(0, inplace=True)
df3['备货单量'].fillna(0, inplace=True)
df3['揽件前退款单量'].fillna(0, inplace=True)
```

```python
#df3.to_excel('包状态.xlsx')
```

```python
# 平均客单涉及的两项都需要注意： 平均客单 = 总发出单量的销售金额/总发出单量
# 复购此前是基于总发出单量,先修正订单数
df3['销售金额'].fillna(0, inplace=True)
df3['订单数'] = df3['总支付单量'] - df3['售前退款单量'] - df3['备货单量']
df3['揽件订单数'] = df3['总支付单量'] - df3['揽件前退款单量'] - df3['备货单量']
for i in range(0,len(df3)):
    if df3.loc[i,'订单数'] == 0:
        df3.loc[i,'销售金额'] = 0
```

```python
df3["已发货"] = 0
df3["在线支付已完成"] = 0
df3["售后中"] = 0
df3["退回成功"] = 0
for i in range(0,len(df3)):
    if df3.loc[i,"订单状态"] == "已完成" and (df3.loc[i,"售后状态"] == "换货成功" or df3.loc[i,"售后状态"] == '-' or df3.loc[i,"售后状态"] == '售后关闭' or df3.loc[i,"售后状态"] == '补寄/换货成功' or df3.loc[i,"售后状态"] == '补寄成功'):
        df3.loc[i,"在线支付已完成"] = df3.loc[i,'订单数']

    elif df3.loc[i,'售后状态'] == '退款成功':
        df3.loc[i,"退回成功"] = df3.loc[i,'订单数']

    elif df3.loc[i,"订单状态"] == '已发货' and (df3.loc[i,"售后状态"] == "换货成功" or df3.loc[i,"售后状态"] == '-' or df3.loc[i,"售后状态"] == '售后关闭' or df3.loc[i,"售后状态"] == '补寄/换货成功' or df3.loc[i,"售后状态"] == '补寄成功'):
        df3.loc[i,"已发货"] = df3.loc[i,'订单数']
        
    elif df3.loc[i,'售后状态'] != "-" and df3.loc[i,'售后状态'] != "换货成功" and df3.loc[i,'售后状态'] != "售后关闭" and df3.loc[i,'售后状态'] != "退款成功":
        df3.loc[i,"售后中"] = df3.loc[i,'订单数']
```

```python
#综合订单表、编码表、异常运单表后做透视表
import numpy as np
df3['销售金额'].fillna(0, inplace=True)
result = pd.pivot_table(df3,index=['订单提交时间'] , values = ['销售金额','订单数','在线支付已完成','售后中','退回成功','已发货','复购','广告','总支付单量','售前退款单量','备货单量','揽件前退款单量','揽件订单数'] , aggfunc = np.sum)
result = result.reset_index()
result = result[['订单提交时间','订单数','销售金额','售后中','在线支付已完成','已发货','退回成功','复购','广告','总支付单量','售前退款单量','备货单量','揽件前退款单量','揽件订单数']]
```

```python
#拼接店铺消耗表
df_consume = pd.read_excel("店铺消耗.xlsx")
df_consume = df_consume[['日期','产品','消耗','转化数']]
temp = df_consume.日期.str.extract('(\d+).*?(\d+).*?(\d+)') 
df_consume['订单提交时间'] = temp[0] + "-" + temp[1] + '-' + temp[2]
del df_consume['日期']
df_consume = pd.pivot_table(df_consume,index=['订单提交时间'] , values = ['消耗','转化数'] , aggfunc = np.sum)
df_consume = df_consume.reset_index()
df_consume["订单提交时间"] = pd.to_datetime(df_consume["订单提交时间"],format = "%Y-%m-%d")
df_consume["订单提交时间"] = df_consume["订单提交时间"].map(lambda x: x.strftime('%Y-%m-%d'))
```

```python
# 销售金额、订单数
final = pd.merge(result,df_consume,on = ['订单提交时间'],how = 'left')
final = final[['订单提交时间','转化数','订单数','在线支付已完成','已发货','售后中','复购','退回成功','消耗','销售金额','总支付单量','售前退款单量','备货单量','揽件前退款单量','揽件订单数']]
final['真实发货率'] = final['订单数'] / final['转化数'] 
final['签收率'] = final['在线支付已完成'] / final['转化数'] 
final['退回率'] = final['退回成功'] / final['转化数'] 
final['转化成本'] = final['消耗'] / final['转化数']
final['平均客单'] = final['销售金额'] / final['订单数']
final['复购率'] = final['复购'] / final['转化数']

final['对标揽件单量签收率'] = final['在线支付已完成'] / final['揽件订单数'] 
final['对标揽件单量退回率'] = final['退回成功'] / final['揽件订单数'] 
final['对标揽件单量复购率'] = final['复购'] / final['揽件订单数'] 
final['售前退款率'] = final['售前退款单量'] / final['总支付单量']
final['揽件前退款率'] = final['揽件前退款单量'] / final['总支付单量']

final['真实发货率'] = final['真实发货率'].apply(lambda x: format(x, '.2%'))
final['签收率'] = final['签收率'].apply(lambda x: format(x, '.2%'))
final['退回率'] = final['退回率'].apply(lambda x: format(x, '.2%'))
final['复购率'] = final['复购率'].apply(lambda x: format(x, '.2%'))
final['对标揽件单量签收率'] = final['对标揽件单量签收率'].apply(lambda x: format(x, '.2%'))
final['对标揽件单量退回率'] = final['对标揽件单量退回率'].apply(lambda x: format(x, '.2%'))
final['对标揽件单量复购率'] = final['对标揽件单量复购率'].apply(lambda x: format(x, '.2%'))
final['售前退款率'] = final['售前退款率'].apply(lambda x: format(x, '.2%'))
final['揽件前退款率'] = final['揽件前退款率'].apply(lambda x: format(x, '.2%'))
final['转化成本'] = final['转化成本'].round(decimals=2)
final['平均客单'] = final['平均客单'].round(decimals=2)

final.rename(columns={"转化数":"广告单量","订单数":"总发出单量","订单提交时间":"日期",'销售金额':'客单'},inplace=True)
final['产品'] = '直播产品'
final= final[['日期','产品','广告单量','总发出单量','真实发货率','总支付单量','售前退款率','揽件前退款率','在线支付已完成','已发货','售后中','退回成功','签收率','退回率','复购率','消耗','对标揽件单量签收率',
             '对标揽件单量退回率','对标揽件单量复购率','平均客单','客单','复购','备货单量','售前退款单量','揽件前退款单量']]
```

```python
# 生成广告表
df_advert = df3[(df3['流量体裁']=='直播')|(df3['流量体裁'] == '商品卡')]
result_advert = pd.pivot_table(df_advert,index=['订单提交时间'] , values = ['销售金额','订单数','在线支付已完成','售后中','退回成功','已发货','复购','广告','总支付单量','售前退款单量','备货单量','揽件前退款单量','揽件订单数'] , aggfunc = np.sum)
final_advert = pd.merge(result_advert,df_consume,on = ['订单提交时间'],how = 'left')
final_advert = final_advert[['订单提交时间','转化数','订单数','在线支付已完成','已发货','售后中','退回成功','消耗','销售金额','复购','广告','总支付单量','售前退款单量','备货单量','揽件前退款单量','揽件订单数']]
final_advert['真实发货率'] = final_advert['订单数'] / final_advert['转化数'] 
final_advert['签收率'] = final_advert['在线支付已完成'] / final_advert['转化数'] 
final_advert['退回率'] = final_advert['退回成功'] / final_advert['转化数'] 
final_advert['转化成本'] = final_advert['消耗'] / final_advert['转化数']
final_advert['平均客单'] = final_advert['销售金额'] / final_advert['订单数']

final_advert['对标揽件单量签收率'] = final_advert['在线支付已完成'] / final_advert['揽件订单数'] 
final_advert['对标揽件单量退回率'] = final_advert['退回成功'] / final_advert['揽件订单数'] 
final_advert['对标揽件单量复购率'] = final_advert['复购'] / final_advert['揽件订单数']
final_advert['售前退款率'] = final_advert['售前退款单量'] / final_advert['总支付单量']
final_advert['揽件前退款率'] = final_advert['揽件前退款单量'] / final_advert['总支付单量']

final_advert['真实发货率'] = final_advert['真实发货率'].apply(lambda x: format(x, '.2%'))
final_advert['签收率'] = final_advert['签收率'].apply(lambda x: format(x, '.2%'))
final_advert['退回率'] = final_advert['退回率'].apply(lambda x: format(x, '.2%'))
final_advert['对标揽件单量签收率'] = final_advert['对标揽件单量签收率'].apply(lambda x: format(x, '.2%'))
final_advert['对标揽件单量退回率'] = final_advert['对标揽件单量退回率'].apply(lambda x: format(x, '.2%'))
final_advert['对标揽件单量复购率'] = final_advert['对标揽件单量复购率'].apply(lambda x: format(x, '.2%'))
final_advert['售前退款率'] = final_advert['售前退款率'].apply(lambda x: format(x, '.2%'))
final_advert['揽件前退款率'] = final_advert['揽件前退款率'].apply(lambda x: format(x, '.2%'))
final_advert['转化成本'] = final_advert['转化成本'].round(decimals=2)
final_advert['平均客单'] = final_advert['平均客单'].round(decimals=2)

final_advert.rename(columns={"转化数":"广告单量","订单数":"总发出单量","订单提交时间":"日期",'销售金额':'客单'},inplace=True)
final_advert['产品'] = '直播产品'
final_advert= final_advert[['日期','产品','广告单量','总发出单量','真实发货率','总支付单量','售前退款率','揽件前退款率','在线支付已完成','已发货','售后中','退回成功','签收率','退回率','消耗','对标揽件单量签收率',
                            '对标揽件单量退回率','平均客单','客单','备货单量','售前退款单量','揽件前退款单量']]
```
