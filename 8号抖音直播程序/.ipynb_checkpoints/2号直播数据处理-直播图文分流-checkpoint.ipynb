```python
import pandas as pd
# df = pd.read_excel("./订单.xlsx")
# df = pd.read_csv("./订单.csv",encoding='gbk')
df =pd.read_csv("./1.csv",encoding='gbk', low_memory=False)
```

```python
# if '\t' in df.loc[0,'商家编码']:
#     df['商家编码'] = df2['商家编码'].str[:19]
if '\t' in df.loc[0,'订单提交时间']:
    df['订单提交时间'] = df['订单提交时间'].str[1:]
```

```python
from datetime import datetime
df['商家编码'] = df['商家编码'].str[1:]
df["订单提交时间"] = pd.to_datetime(df["订单提交时间"],format = "%Y-%m-%d %H:%M:%S")
df["订单提交时间"] = df["订单提交时间"].map(lambda x: x.strftime('%Y-%m-%d'))
df["销售金额"]= (df['商品数量'] * df['商品单价']) + (df['运费'])
df1 = pd.read_excel("商品编码.xlsx")
df1.rename(columns={"商品名":"产品名称","商品编码":"商家编码"},inplace=True)
df1 = df1[["商家编码","产品名称"]]
df= pd.merge(df,df1,on="商家编码",how="left")
df = df.reset_index(drop=True)
```

```python
# 缺码表
lack_code = df[df['产品名称'].isnull()]['商家编码'].value_counts()
lack_code.to_excel('缺码表.xlsx')
```

```python
df2 = pd.read_csv("./2.csv",encoding='utf-8')
#df2 = pd.read_csv("./2.csv",encoding='gbk')
# df2 = pd.read_csv("./3.csv")
# df2 = pd.read_csv("./异常运单数.csv",encoding='gbk')
# df2 = pd.read_excel("./异常运单数.xlsx")
df2.rename(columns={"订单号":"主订单编号"},inplace= True)
df['主订单编号'] = df['主订单编号'].astype(str)
df2['主订单编号'] = df2['主订单编号'].astype(str)
df2 = df2[["主订单编号","包裹状态","揽件时间"]]
df2["包裹状态"].replace('已签收','已完成1',inplace= True)
# df2["包裹状态"].replace('已拒收','已完成1',inplace= True)
df2["包裹状态"].replace('派送中','已完成1',inplace= True)
df2["包裹状态"].replace('待取件','已完成1',inplace= True)
df2 = df2.drop_duplicates(subset='主订单编号', keep='first') #删除重复项
if '\t' in df2.loc[0,'主订单编号']:
    df2['主订单编号'] = df2['主订单编号'].str[:19]
```

```python
df3 = pd.merge(df,df2,on="主订单编号",how="left")
df3['主订单编号'] = df3['主订单编号'].astype(str)
df3['子订单编号'] = df3['子订单编号'].astype(str)
# df3 = df3.dropna(subset=['包裹状态'])
# df3.reset_index(drop=True,inplace= True)
for i in range (0,len(df3)):
    if df3.loc[i,"包裹状态"] == "已完成1" and (df3.loc[i,'售后状态'] == '-' or df3.loc[i,'售后状态'] == '售后关闭' or df3.loc[i,'售后状态'] == '换货成功'):
        df3.loc[i,"订单状态"] = "已完成"
```

```python
# 订单数会rename为总发出单量，stream = 直播， = 图文；
df3['直播订单数'] = df3['商品数量']
df3['图文订单数'] = 1
```

```python
# 需要清除未支付订单：①取消原因：非空②包裹状态：空 并且 订单状态：关闭 并且 售后状态：-
# 排除 (订单状态：已关闭 且 售后状态：- ) 即可实现前一点
df3.drop(df3[(df3.订单状态 == '已关闭') & (df3.售后状态 == '-')].index, inplace=True) 
df3.reset_index(drop=True,inplace= True)
```

```python
df3[df3['包裹状态'].isnull()]['订单状态'].value_counts()
```

```python
# 直播
df3['直播总支付单量'] = df3['直播订单数']
df3['直播售前退款单量'] = df3[(df3['包裹状态'].isnull())&(df3['售后状态']=='退款成功')]['直播订单数']
df3['直播备货单量'] = df3[(df3['包裹状态'].isnull())&(df3['订单状态']=='待发货')]['直播订单数']
df3['直播售前退款单量'].fillna(0, inplace=True)
df3['直播备货单量'].fillna(0, inplace=True)
```

```python
# 图文
df3['图文总支付单量'] = df3['图文订单数']
df3['图文售前退款单量'] = df3[(df3['包裹状态'].isnull())&(df3['售后状态']=='退款成功')]['图文订单数']
df3['图文备货单量'] = df3[(df3['包裹状态'].isnull())&(df3['订单状态']=='待发货')]['图文订单数']
df3['图文售前退款单量'].fillna(0, inplace=True)
df3['图文备货单量'].fillna(0, inplace=True)
```

```python
# 平均客单涉及的两项都需要注意： 平均客单 = 总发出单量的销售金额/总发出单量
# 复购此前是基于总发出单量,先修正订单数
df3['销售金额'].fillna(0, inplace=True)
df3['直播订单数'] = df3['直播总支付单量'] - df3['直播售前退款单量'] - df3['直播备货单量']
df3['图文订单数'] = df3['图文总支付单量'] - df3['图文售前退款单量'] - df3['图文备货单量']
for i in range(0,len(df3)):
    if df3.loc[i,'直播订单数'] == 0 or df3.loc[i,'图文订单数'] == 0:
        df3.loc[i,'销售金额'] = 0
```

```python
#df3.to_excel("检查.xlsx",index=False)
```

```python
df3["直播已发货"] = 0
df3["直播在线支付已完成"] = 0
df3["直播售后中"] = 0
df3["直播退回成功"] = 0
df3["图文已发货"] = 0
df3["图文在线支付已完成"] = 0
df3["图文售后中"] = 0
df3["图文退回成功"] = 0
# 标记订单状态：方便复购分状态分流
df3["标签"] = "" 
for i in range(0,len(df3)):
    if df3.loc[i,"订单状态"] == "已完成" and (df3.loc[i,"售后状态"] == "换货成功" or df3.loc[i,"售后状态"] == '-' or df3.loc[i,"售后状态"] == '售后关闭' or df3.loc[i,"售后状态"] == '补寄/换货成功' or df3.loc[i,"售后状态"] == '补寄成功'):
        df3.loc[i,"直播在线支付已完成"] = df3.loc[i,'直播订单数']
        df3.loc[i,"图文在线支付已完成"] = df3.loc[i,'图文订单数']
        df3.loc[i,"标签"] = "在线支付已完成"

    elif df3.loc[i,'售后状态'] == '退款成功' or df3.loc[i,'售后状态'] == '退款中':
        df3.loc[i,"直播退回成功"] = df3.loc[i,'直播订单数']
        df3.loc[i,"图文退回成功"] = df3.loc[i,'图文订单数']
        df3.loc[i,"标签"] = "退款成功"
    
    elif df3.loc[i,"订单状态"] == '已发货' and (df3.loc[i,"售后状态"] == "换货成功" or df3.loc[i,"售后状态"] == '-' or df3.loc[i,"售后状态"] == '售后关闭' or df3.loc[i,"售后状态"] == '补寄/换货成功' or df3.loc[i,"售后状态"] == '补寄成功'):
        df3.loc[i,"直播已发货"] = df3.loc[i,'直播订单数']
        df3.loc[i,"图文已发货"] = df3.loc[i,'图文订单数']
        df3.loc[i,"标签"] = "已发货"
        
    elif df3.loc[i,'售后状态'] != "-" and df3.loc[i,'售后状态'] != "换货成功" and df3.loc[i,'售后状态'] != "售后关闭" and df3.loc[i,'售后状态'] != "退款成功" and df3.loc[i,'售后状态'] != '退款中':
        df3.loc[i,"直播售后中"] = df3.loc[i,'直播订单数']
        df3.loc[i,"图文售后中"] = df3.loc[i,'图文订单数']
        df3.loc[i,"标签"] = "售后中"
```

```python
# 复购按比例分流
ad = df3[(df3['广告渠道']=='短视频')|(df3['广告渠道']=='直播')] # 广告：直播+短视频
repeat = df3[(df3['广告渠道']=='无')|(df3['广告渠道']=='商品卡')] # 非广告：复购+自然流
stream = ad[ad['广告渠道']=='直播'] # 直播
short_video = ad[ad['广告渠道']=='短视频'] # 短视频

# tags = ['在线支付已完成','退款成功','已发货','售后中']
tags = df3['标签'].unique().tolist() 
# 直播占比：直播在广告订单中的占比
stream_percent = ad['广告渠道'].value_counts(normalize=True).直播

for tag in tags:
    # 按直播占比，随机选取复购单量
    data = repeat[repeat['标签']==tag]
    stream_repeat = data.sample(int(stream_percent*len(data)))
    stream_repeat_index = stream_repeat.index
    # 其余归入图文复购
    repeat_index = data.index
    short_video_repeat_index = repeat_index.difference(stream_repeat_index)
    short_video_repeat = data.loc[short_video_repeat_index]
    # 保存
    stream = pd.concat([stream, stream_repeat], ignore_index=True)
    short_video = pd.concat([short_video, short_video_repeat], ignore_index=True)
```

```python
stream_percent
```

```python
# 直播
stream['复购'] = stream[(stream['广告渠道']== '无')|(stream['广告渠道']== '商品卡')]['直播订单数']
stream['复购'].fillna(0, inplace=True)
# 图文
short_video['复购'] = short_video[(short_video['广告渠道']== '无')|(short_video['广告渠道']== '商品卡')]['图文订单数']
short_video['复购'].fillna(0, inplace=True)
```

```python
#综合订单表、编码表、异常运单表后做透视表-直播
import numpy as np
stream['销售金额'].fillna(0, inplace=True)
stream_result = pd.pivot_table(stream,index=['订单提交时间'] , values = ['销售金额','直播订单数','直播在线支付已完成','直播售后中','直播退回成功','直播已发货','复购','直播总支付单量','直播售前退款单量','直播备货单量'] , aggfunc = np.sum)
stream_result = stream_result.reset_index()
stream_result = stream_result[['订单提交时间','直播订单数','销售金额','直播售后中','直播在线支付已完成','直播已发货','直播退回成功','复购','直播总支付单量','直播售前退款单量','直播备货单量']]
```

```python
#店铺消耗表
df_consume = pd.read_excel("店铺消耗.xlsx")
```

```python
#拼接店铺消耗表—直播
stream_consume = df_consume[df_consume['备注'].str.contains('直播')]
stream_consume = stream_consume[['日期','产品','消耗','转化数']]
temp = stream_consume.日期.str.extract('(\d+).*?(\d+).*?(\d+)') 
stream_consume['订单提交时间'] = temp[0] + "-" + temp[1] + '-' + temp[2]
del stream_consume['日期']
stream_consume = pd.pivot_table(stream_consume,index=['订单提交时间'] , values = ['消耗','转化数'] , aggfunc = np.sum)
stream_consume = stream_consume.reset_index()
stream_consume["订单提交时间"] = pd.to_datetime(stream_consume["订单提交时间"],format = "%Y-%m-%d")
stream_consume["订单提交时间"] = stream_consume["订单提交时间"].map(lambda x: x.strftime('%Y-%m-%d'))
```

```python
# 汇总表--直播
final = pd.merge(stream_result,stream_consume,on = ['订单提交时间'],how = 'left')
final = final[['订单提交时间','转化数','直播订单数','直播在线支付已完成','直播已发货','直播售后中','复购','直播退回成功','消耗','销售金额','直播总支付单量','直播售前退款单量','直播备货单量']]
final['真实发货率'] = final['直播订单数'] / final['转化数'] 
final['签收率'] = final['直播在线支付已完成'] / final['转化数'] 
final['退回率'] = final['直播退回成功'] / final['转化数'] 
final['转化成本'] = final['消耗'] / final['转化数']
final['平均客单'] = final['销售金额'] / final['直播订单数']
final['复购率'] = final['复购'] / final['转化数']
final['对标总发出单量签收率'] = final['直播在线支付已完成'] / final['直播订单数'] 
final['对标总发出单量退回率'] = final['直播退回成功'] / final['直播订单数'] 
final['对标总发出单量复购率'] = final['复购'] / final['直播订单数'] 
final['售前退款率'] = final['直播售前退款单量'] / final['直播总支付单量']
final['真实发货率'] = final['真实发货率'].apply(lambda x: format(x, '.2%'))
final['签收率'] = final['签收率'].apply(lambda x: format(x, '.2%'))
final['退回率'] = final['退回率'].apply(lambda x: format(x, '.2%'))
final['复购率'] = final['复购率'].apply(lambda x: format(x, '.2%'))
final['对标总发出单量签收率'] = final['对标总发出单量签收率'].apply(lambda x: format(x, '.2%'))
final['对标总发出单量退回率'] = final['对标总发出单量退回率'].apply(lambda x: format(x, '.2%'))
final['对标总发出单量复购率'] = final['对标总发出单量复购率'].apply(lambda x: format(x, '.2%'))
final['售前退款率'] = final['售前退款率'].apply(lambda x: format(x, '.2%'))
final['转化成本'] = final['转化成本'].round(decimals=2)
final['平均客单'] = final['平均客单'].round(decimals=2)
final.rename(columns={"转化数":"广告单量","直播订单数":"总发出单量","订单提交时间":"日期",'销售金额':'客单'},inplace=True)
final['产品'] = '直播产品'
final= final[['日期','产品','广告单量','总发出单量','真实发货率','直播总支付单量','直播售前退款单量','售前退款率','直播在线支付已完成','直播已发货','直播售后中','直播退回成功','签收率','退回率','复购率','消耗','对标总发出单量签收率',
             '对标总发出单量退回率','对标总发出单量复购率','平均客单','客单','复购','直播备货单量']]
```
